```python
from collections import deque

graph = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['G'],
    'C': ['E'],
    'D': ['G'],
    'E': ['G']
}

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = {start}

    while queue:

        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        for neighbour in graph[node]:

            if neighbour not in visited:

                visited.add(neighbour)
                queue.append(path + [neighbour])

    return None


result = bfs(graph, 'S', 'G')

print("Path:", result)
```

```python
import heapq

def ucs(graph, start, goal):

    priority_queue = [(0, start, [start])]
    visited = {}

    while priority_queue:

        cost, node, path = heapq.heappop(priority_queue)

        if node == goal:
            return path, cost

        if node in visited and visited[node] <= cost:
            continue

        visited[node] = cost

        for neighbour, weight in graph[node]:

            new_cost = cost + weight

            if neighbour not in visited or new_cost < visited[neighbour]:
                heapq.heappush(
                    priority_queue,
                    (new_cost, neighbour, path + [neighbour])
                )

    return None


weighted_graph = {
    'A': [('B', 2), ('C', 1)],
    'B': [('G', 10)],
    'C': [('G', 2)],
    'G': []
}

print(ucs(weighted_graph, 'A', 'G'))
```

```python

graph = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['G'],
    'C': ['E'],
    'D': ['G'],
    'E': ['G']
}

def depth_limited_search(graph, node, goal, limit):

    if node == goal:
        return [node]

    if limit == 0:
        return None

    for neighbour in graph[node]:

        result = depth_limited_search(
            graph, neighbour, goal, limit - 1
        )

        if result is not None:
            return [node] + result

    return None


def iterative_deepening(graph, start, goal, max_depth):

    for depth in range(max_depth + 1):

        result = depth_limited_search(
            graph, start, goal, depth
        )

        if result is not None:
            return result

    return None


print(iterative_deepening(graph, 'S', 'G', 10))
```

```python
game_tree = {
    "A": {
        "A1": [3, 5],
        "A2": [2, 9]
    }
}

# Evaluate MIN nodes
left_value = min(game_tree["A"]["A1"])
right_value = min(game_tree["A"]["A2"])

# Evaluate MAX root
best_value = max(left_value, right_value)

print("Left MIN value:", left_value)
print("Right MIN value:", right_value)
print("Best MAX value:", best_value)
```
