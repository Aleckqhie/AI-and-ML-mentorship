```python
def group_event_by_hour(events):
    hour_buckets ={}
    for event in events :
        timestamp =event['ts']
        event_type=event['type']
        hour_prefix=timestamp[0:14]
        if hour_prefix not in hour_buckets:
            hour_buckets[hour_prefix]=[]
            hour_buckets[hour_prefix].append(event_type)
    return hour_buckets
group_event_by_hour([
  {"ts":"2024-01-15T14:32:00","type":"click"},
  {"ts":"2024-01-15T14:55:00","type":"view"},
  {"ts":"2024-01-15T15:01:00","type":"click"}
])
```
